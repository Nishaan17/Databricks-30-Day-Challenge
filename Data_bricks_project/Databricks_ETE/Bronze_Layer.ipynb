```python
df = spark.read.format('parquet').load('abfss://source@datalakeete1.dfs.core.windows.net/orders')
df.display()
```

While reading we should get the schema location, so that it can wirte it there in bronze storage.
Everything will be in memory, it wont execute while reading

```python
df = spark.readStream.format('cloudFiles')\
        .option('cloudFiles.format','parquet')\
        .option('cloudFiles.schemaLocation','abfss://bronze@datalakeete1.dfs.core.windows.net/checkpoint_orders')\
        .load('abfss://source@datalakeete1.dfs.core.windows.net/orders')
```

After reading, when writing , we need to load the checkpointLocation , so that databricks get to know, which files has been processed.

```python
df.writeStream.format('parquet')\
    .outputMode('append')\
    .option('checkpointLocation','abfss://bronze@datalakeete1.dfs.core.windows.net/checkpoint_orders')\
    .option('path','abfss://bronze@datalakeete1.dfs.core.windows.net/orders')\
    .trigger(once=True)\
    .start()    
    # .trigger(processingTime='10 seconds')\
    
```
