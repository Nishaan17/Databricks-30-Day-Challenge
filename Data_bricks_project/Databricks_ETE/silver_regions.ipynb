### Reading delta table

```python
df =spark.read.table('databricks_cat_1234.bronze.regions')
df.display()
```

```python
df=df.drop('_rescued_data')
```

```python
df.display()
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save("abfss://silver@datalakeete1.dfs.core.windows.net/regions")
```

```python
df.printSchema()
```
