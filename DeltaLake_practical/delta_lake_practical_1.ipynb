# Manged vs external table

## Manged table

```python
%sql
Create database sales;
```

```python
%sql
CReate table sales.mantable(
    id int,
    name string,
    mark int
) using Delta;
```

```python
%sql
INSERT into sales.mantable values(1,'abc',10),(2,'def',20),(3,'ghi',30),(4,'jkl',40);
```

```python
%sql
select * from sales.mantable;
```

If we drop, all the data will be erased from meta store as well from storage(by data bricks, whcih we give when starting databricks as managed workspace)

## External Delta table

```python
%sql
CReate table If not exists sales.exttable(
    id int,
    name string,
    mark int
) using Delta
location 'abfss://destination@datalakenishaan.dfs.core.windows.net/Salesdb/exttable';
```

```python
%sql
INSERT into sales.exttable values(5,'abc',10),(6,'def',20),(7,'ghi',30),(8,'jkl',40);
```

```python
%sql
select * from sales.exttable;
```

```python
%sql
drop table sales.exttable;
```

```python
%sql
select * from sales.exttable;
```
