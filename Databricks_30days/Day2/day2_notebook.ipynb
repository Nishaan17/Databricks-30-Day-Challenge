What is Compute?

Compute is the execution environment (data plane) where the code in a Databricks notebook is actually executed. It consists of a Driver node and one or more Worker nodes that run Apache Spark applications.

What is a Driver?

The Driver node is responsible for managing the Spark application. It receives the code from the notebook, creates an execution plan, divides the work into tasks, assigns those tasks to the Worker nodes, and collects the results after processing.


What are Worker Nodes?

Worker nodes execute the tasks assigned by the Driver node. They process the data in parallel and return the results to the Driver node. Multiple Worker nodes allow Spark to process large datasets efficiently.

What is Databricks Runtime?

Databricks Runtime is an optimized runtime environment provided by Databricks. It includes Apache Spark along with additional libraries, Delta Lake support, performance optimizations, and security enhancements. Instead of manually installing and configuring Spark and its dependencies, Databricks Runtime provides everything as a pre-configured environment, making cluster management much easier.


What is Photon?

Photon is Databricks' high-performance query execution engine. It runs on the Worker nodes and executes supported SQL and DataFrame operations using a highly optimized columnar engine written in C++. Compared to the standard Spark engine, Photon can significantly improve the performance of analytical workloads.



Why is the Control Plane separated from the Compute Plane?

The Control Plane is managed by Databricks and is responsible for services such as the workspace UI, notebooks, job scheduling, and cluster management. It does not directly process customer data.

The Compute Plane is where Spark applications execute. It contains the Driver and Worker nodes and is deployed in the customer's cloud environment (Azure, AWS, or GCP), allowing the customer to retain control over their data.

"This is the core of Databricks."

I would avoid that.

A better way to think about it is:

Apache Spark → Distributed processing engine
Databricks Runtime → Optimized Spark distribution from Databricks
Photon → Faster execution engine for supported workloads
Delta Lake → Storage layer with ACID transactions and other reliability features

These are all parts of the Databricks platform, but none of them alone is "the core."
