```python
%sql
create table smart_phones(
    id int,
    name string
);
```

```python
%sql
insert into smart_phones values(1,'apple'),(2,'samsung'),(3,'apple');
```

## Stored view

```python
%sql
create view v_apple as select * from smart_phones where name= 'apple';
```

## Temp view

```python
%sql
create temporary view v_tmp_apple as select * from smart_phones where name= 'apple';
```

## Global Temp view

```python
%sql
create global temporary view v_global_tmp_apple as select * from smart_phones where name= 'apple';
```
