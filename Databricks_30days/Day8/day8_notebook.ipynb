```python
from pyspark.sql import SparkSession

employees = spark.createDataFrame([
    (1, "John", 10),
    (2, "Alice", 20),
    (3, "Bob", 30),
    (4, "David", 40),
    (5, "Eva", None)
], ["emp_id", "name", "dept_id"])

departments = spark.createDataFrame([
    (10, "IT"),
    (20, "HR"),
    (30, "Finance"),
    (50, "Marketing")
], ["dept_id", "department"])

employees.show()
departments.show()
```

Inner Join

```python
inner_df = employees.join(departments, "dept_id", "inner")

inner_df.show()

inner_df.explain("formatted")
```

Left Join

```python
left_df = employees.join(departments, "dept_id", "left")

left_df.show()

left_df.explain("formatted")
```

Right Join

```python
right_df = employees.join(departments, "dept_id", "right")

right_df.show()

right_df.explain("formatted")
```

Full Join

```python
full_df = employees.join(departments, "dept_id", "full")

full_df.show()

full_df.explain("formatted")
```

Left Semi

```python
semi_df = employees.join(
    departments,
    "dept_id",
    "left_semi"
)

semi_df.show()

semi_df.explain("formatted")
```

Left Anti Join

```python
anti_df = employees.join(
    departments,
    "dept_id",
    "left_anti"
)

anti_df.show()

anti_df.explain("formatted")
```

Force a Broadcast Join

```python
from pyspark.sql.functions import broadcast

broadcast_df = employees.join(
    broadcast(departments),
    "dept_id"
)

broadcast_df.explain("formatted")
```
