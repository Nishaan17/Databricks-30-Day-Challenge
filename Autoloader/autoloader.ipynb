# Autoloader

cloudFiles.format tells Auto Loader what type of files it should ingest (CSV, JSON, Parquet, XML, etc.).
cloudFiles.schemaLocation is the location where Auto Loader stores the inferred schema and schema history.

On the first run, Auto Loader infers the schema from the source files and saves it here.

On subsequent runs, it reads the stored schema instead of inferring it again, which improves performance and enables schema evolution.


It is a folder where Auto Loader automatically stores and manages the schema.



```python
df = spark.readStream.format('cloudFiles')\
    .option('cloudFiles.format', 'csv')\
    .option('cloudFiles.schemaLocation', 'abfss://aldestination@datalakenishaan.dfs.core.windows.net/schema')\
    .load('abfss://alsource@datalakenishaan.dfs.core.windows.net/')
```

Checkpoint stores the streaming query state, including:

which files have already been processed,
streaming offsets,
commit information,
metadata needed to recover the stream after a restart.




                Source Folder
                      │
          customers1.csv
          customers2.csv
          customers3.csv
                      │
                      ▼
              Auto Loader
          (cloudFiles.format)
                      │
        Infer schema (first run)
                      │
                      ▼
             schemaLocation
      (stores inferred schema)
                      │
                      ▼
           Structured Streaming
                      │
      checkpointLocation
 (stores processed files, offsets,
      commits, stream state)
                      │
                      ▼
               Delta Table

```python
df.writeStream \
    .format("delta") \
    .option(
        "checkpointLocation",
        "abfss://aldestination@datalakenishaan.dfs.core.windows.net/checkpoint"
    ) \
    .trigger(availableNow=True) \
    .start("abfss://aldestination@datalakenishaan.dfs.core.windows.net/data12")
```
